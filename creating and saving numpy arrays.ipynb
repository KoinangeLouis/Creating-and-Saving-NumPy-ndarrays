unlike python lists, numpy's elements must have the same dtyoe.

```python
f = np.array([1, 1.27, 7])
```

```python
print(f)
print("shape:", f.shape)
print("type:", type(f))
print("dtype:", f.dtype)
```

since all the elements of the numpy array must be of the same type, numpy upcasts all integers to float to avoid losing precision.

```python
k = np.array([1.4, 5.8, 7.9], dtype=np.int64)
```

```python
print(k)
print("dtype:", k.dtype)
```
